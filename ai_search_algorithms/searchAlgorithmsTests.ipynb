### Romênia, Arad-> Bucarest

```python
# Crio a estrutura problema nos formato do livro.

# Importo os dados brutos do problema
path = Path("romenia.json")
with path.open("r", encoding="utf-8") as f:
    romenia = json.load(f)
    
problem = {}
problem['states'] = list(romenia['edges'].keys())
problem['initial_state'] = romenia['initial_state']
problem['goal_state'] = romenia['goal']
problem['actions'] = {state: ['to'+s for s in romenia['edges'][state].keys()] for state in problem['states']}
problem['result'] = {(state,s): s[2:] for state in problem['actions'].keys() for s in problem['actions'][state]}
problem['action_cost'] = {(s,a,problem['result'][(s,a)]): romenia['edges'][s][problem['result'][(s,a)]] for s in romenia['edges'] for a in problem['actions'][s]}
problem['heuristic'] = romenia['h']
```

```python
result = best_first_search(problem, 'bfs')
result
```

```python
result = best_first_search(problem, 'bfs', )
result
```

```python
rslt = best_first_search(problem, 'gbf')
rslt
```

```python
result = best_first_search(problem, 'uniform')
result
```

```python
rslt = best_first_search(problem, 'a*')
rslt
```
